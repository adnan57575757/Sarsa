##SARSA

```python
gamma = 0.99
alpha = 0.5
epsilon = 0.1 # 10% explore and 90% exploit
episodes = 500
```

```python
# Q-table = store 0-values ,size =48*4

Q = np.zeros((48,4))
Q
```

```python
#policy -epsilon-greedy : state -> action]

def epsilon_greedy(state):
  if random.random() < epsilon:  #generate a random no.  and compare with epsilon
    return env.action_space.sample()   # random action => EXPLORE
  else:
    return np.argmax(Q[state]) # exploit
```

```python
for episode in range(episodes):

  env = gym.make("CliffWalking-v1")

  done = False #when the episode has ended - reached destination
  state, _ = env.reset() #current state
  action = epsilon_greedy(state) #current action

  total_reward = 0
  episode_len = 0

  while not done:
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated #terminated -reached destination, trucated - exceeds no. of path length
    next_action = epsilon_greedy(next_state)

    #SARSA update
    Q[state, action] += alpha * (reward + gamma*(Q[next_state, next_action] - Q[state, action]))

    state = next_state
    action = next_action

    total_reward += reward
    episode_len += 1

  print(f"episode{episode+1}/500 tot reward: {total_reward} & ep len : {episode_len}")
  env.close()






```

```python
# wht did our agent learn?

env = gym.make("CliffWalking-v1", render_mode="human")
state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state]) #No training only action
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated



    total_reward += reward
    episode_len += 1

print(f"tot reward= {total_reward} & ep len : {episode_len}")
env.close()
```
